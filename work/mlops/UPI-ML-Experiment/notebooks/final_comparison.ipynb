```python
import pandas as pd
import matplotlib.pyplot as plt

# Load final ensemble results
dataset1 = pd.read_csv("../results/dataset1_ensemble_results.csv")
dataset2 = pd.read_csv("../results/dataset2_ensemble_results.csv")
dataset3 = pd.read_csv("../results/dataset3_ensemble_results.csv")

print("All final ensemble results loaded successfully!")

print("Dataset 1:", dataset1.shape)
print("Dataset 2:", dataset2.shape)
print("Dataset 3:", dataset3.shape)
```

```python
# ============================================
# FINAL DATASET COMPARISON
# ============================================

final_results = pd.DataFrame({
    "Dataset": [
        "Dataset 1",
        "Dataset 2",
        "Dataset 3"
    ],

    "Total Records": [
        len(dataset1),
        len(dataset2),
        len(dataset3)
    ],

    "Final Anomalies": [
        dataset1["final_anomaly"].sum(),
        dataset2["final_anomaly"].sum(),
        dataset3["final_anomaly"].sum()
    ]
})

final_results["Anomaly Percentage"] = (
    final_results["Final Anomalies"]
    / final_results["Total Records"]
    * 100
)

display(final_results)
```

```python
# ============================================
# ANOMALY PERCENTAGE COMPARISON
# ============================================

plt.figure(figsize=(8, 5))

plt.bar(
    final_results["Dataset"],
    final_results["Anomaly Percentage"]
)

plt.xlabel("Dataset")
plt.ylabel("Anomaly Percentage (%)")
plt.title("Final Anomaly Percentage Across Datasets")

plt.tight_layout()
plt.show()
```

```python
# ============================================
# FINAL ANOMALY COUNT COMPARISON
# ============================================

plt.figure(figsize=(8, 5))

plt.bar(
    final_results["Dataset"],
    final_results["Final Anomalies"]
)

plt.xlabel("Dataset")
plt.ylabel("Number of Anomalies")
plt.title("Final Anomaly Count Across Datasets")

plt.tight_layout()
plt.show()
```
